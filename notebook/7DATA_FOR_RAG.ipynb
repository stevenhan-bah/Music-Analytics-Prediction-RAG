Clean release_name

```python
df = df[~df['release_name'].str.startswith('[')]
```

```python
from pypinyin import pinyin, Style

# Convert Chinese characters to Pinyin
df['release_name'] = df['release_name'].apply(lambda x: ''.join([p[0] for p in pinyin(x, style=Style.NORMAL)]) if isinstance(x, str) else x)
```

```python
from korean_romanizer.romanizer import Romanizer

def korean_to_romanized(text):
    """
    Convert Korean Hangul to Romanized Latin text.
    """
    if isinstance(text, str):
        return Romanizer(text).romanize()
    return text

# Convert Korean artist names or song names
df['release_name'] = df['release_name'].apply(korean_to_romanized)
```

```python
import pykakasi

kakasi = pykakasi.kakasi()

def japanese_to_romanized(text):
    """
    Convert Japanese text (Kanji, Hiragana, Katakana) into Romanized Latin text.
    """
    if isinstance(text, str):
        result = kakasi.convert(text)
        return ''.join([item['hepburn'] for item in result])
    return text

# Convert Japanese artist or song names
df['release_name'] = df['release_name'].apply(japanese_to_romanized)
```

```python
df['release_name'] = df['release_name'].str.lower()
```

```python
import unicodedata

def remove_diacritics(text):
    """
    Normalize characters with diacritics (e.g., ñ, é) by converting them to basic Latin equivalents (e.g., n, e).
    """
    if isinstance(text, str):
        text = unicodedata.normalize('NFKD', text)  # Normalize to decomposed Unicode (NFKD form)
        return ''.join(c for c in text if not unicodedata.combining(c))  # Remove combining marks
    return text

# Apply the function to clean artist names and other text fields
df['release_name'] = df['release_name'].apply(remove_diacritics)
```

```python
import re

# Remove non-alphanumeric characters
def clean_text(text):
    if isinstance(text, str):
        text = re.sub(r'[^a-zA-Z0-9\s]', '', text)  # Keep only alphanumeric and spaces
        return text.strip()
    return text

# Apply the cleaning to text-based features
df['release_name'] = df['release_name'].apply(clean_text)
```
